233 preserved comparison conversations plus 60 additional conversations (two per prepared scenario). Both versions use the same full JSON prompt and generation settings.
```python
import base64,json,lzma,os,sys,subprocess,hashlib,zipfile
from pathlib import Path
WORK=Path('/kaggle/working/beacon_comparison');WORK.mkdir(exist_ok=True);os.chdir(WORK)
payload=json.loads(lzma.decompress(base64.b64decode('')))
for name,content in payload.items():
 p=WORK/name;p.parent.mkdir(parents=True,exist_ok=True);p.write_text(content,encoding='utf-8')
print('Evaluation files ready. No training.')
```

```python
from huggingface_hub import snapshot_download
repo='abidansari5594/beacon-qualification-qwen2.5-1.5b-qlora'
versions=[('slm/runs/kaggle-v4-continued/adapter', '723dce76fb622f692c7d53658ed52502ec927a72', 'f1d618ed1152e2cb8443bda4e3316e6f5a704c4cb3fa6e8160bbf5e3afc1d99f'), ('slm/runs/kaggle/adapter', '11f9e096fa656f8d26706ce59debb40fc2e2d904', 'e23139ba1b61a4912f476120cef8ff6edb8a8a726a0d8c9a9537308cc492ffe3')]
for target,revision,expected in versions:
 snapshot_download(repo,revision=revision,local_dir=target,allow_patterns=['adapter*','tokenizer*','special_tokens_map.json','added_tokens.json','vocab.json','merges.txt','chat_template.jinja'])
 assert hashlib.sha256((Path(target)/'adapter_model.safetensors').read_bytes()).hexdigest()==expected
 print('Verified version',revision)
```

```python
subprocess.run([sys.executable,'-u','slm/compare_v4.py','--cloud','--batch-size','8'],check=True)
```

```python
folder=WORK/'outputs/beacon-training-v4/comparison'
assert len((folder/'new.jsonl').read_text().splitlines())==293
assert len((folder/'old.jsonl').read_text().splitlines())==60
with zipfile.ZipFile('/kaggle/working/beacon_v4_comparison.zip','w',zipfile.ZIP_DEFLATED) as z:
 for p in folder.iterdir():
  if p.is_file():z.write(p,p.name)
print('COMPARISON COMPLETE. Download beacon_v4_comparison.zip')
```
